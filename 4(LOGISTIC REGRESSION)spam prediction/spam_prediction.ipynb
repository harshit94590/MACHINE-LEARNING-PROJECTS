SPLITTING OUR DATA INTO TRAIN AND TEST

```python
X_train, X_test, y_train, y_test = train_test_split(df["text"], df["label"], test_size=0.2, random_state=42)
```

CREATING A PIPELINE FOR MODEL 

```python
pipeline = Pipeline([("vectorized",CountVectorizer()),("model",MultinomialNB(alpha=5))])
```

TRAINING THE MODEL 

```python
pipeline.fit(X_train,y_train)
```

```python
y_train_pred = pipeline.predict(X_train)
```

```python
pred = pipeline.predict(X_test)
```

```python
print("accuracy_score",accuracy_score(y_test,pred))
print("accuracy_score",accuracy_score(y_train,y_train_pred))
```

SETTING  THE THRESHOLD VALUE TO MINIMIZE FALSE POSITIVE

```python
def spam_clf(email,threshold =0.9):
    probabilities = pipeline.predict_proba(email)
    for i in probabilities[:,1]:
        if  (i > threshold).astype(int):
         print("this email is spam")
        else:
           print("this email is not spam")
```
